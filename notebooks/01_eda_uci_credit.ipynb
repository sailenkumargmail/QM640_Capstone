# 01 — Exploratory Data Analysis: UCI Credit-Card Default

Loads the UCI "default of credit card clients" dataset (real data if prepared via `scripts/prepare_uci_credit.py`, otherwise a schema-accurate synthetic fallback — see `src/dac/data/synthetic.py`) and runs the EDA suite in `src/dac/eda/eda_report.py`: shape/dtypes, missingness, target balance, numeric distributions, correlation with target, and the target rate by `SEX` / `AGE_GROUP` (the protected attributes carried into the fairness audit in notebook 04).

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent / "src"))

from dac.config import CONFIG
from dac.data.loader import load_uci_credit
from dac.eda.eda_report import run_eda

df, is_synthetic = load_uci_credit()
print(f"Loaded {df.shape[0]:,} rows x {df.shape[1]} cols | synthetic={is_synthetic}")
df.head()
```

```python
df.describe(include="all").T
```

```python
uci_cfg = CONFIG["data"]["uci_credit"]
stats = run_eda(
    df,
    target_col=uci_cfg["target_col"],
    protected_attributes=uci_cfg["protected_attributes"],
    figures_dir=CONFIG["paths"]["figures_dir"],
    report_path=CONFIG["paths"]["metrics_dir"] / "eda_uci_credit.md",
    dataset_name="uci_credit",
)
stats
```

See `reports/figures/uci_credit/*.png` for all generated plots (target distribution, missingness, numeric distributions, correlation heatmap, target rate by `SEX` / `AGE_GROUP`).
