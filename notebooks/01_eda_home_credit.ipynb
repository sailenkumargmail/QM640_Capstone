# 01 — Exploratory Data Analysis: Home Credit Default Risk

Loads the Home Credit dataset (real data if downloaded via `scripts/download_home_credit.py`, otherwise a schema-accurate synthetic fallback — see `src/dac/data/synthetic.py`) and runs the EDA suite in `src/dac/eda/eda_report.py`: shape/dtypes, missingness, target balance, numeric distributions, correlation with target, and default-rate breakdown by protected attribute (gender, age group).

Figures are written to `reports/figures/home_credit/` and a markdown summary to `reports/model_performance/eda_home_credit.md`.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent / "src"))

from dac.config import CONFIG
from dac.data.loader import load_home_credit
from dac.eda.eda_report import run_eda

df, is_synthetic = load_home_credit()
print(f"Loaded {df.shape[0]:,} rows x {df.shape[1]} cols | synthetic={is_synthetic}")
df.head()
```

```python
df.describe(include="all").T
```

```python
hc_cfg = CONFIG["data"]["home_credit"]
stats = run_eda(
    df,
    target_col=hc_cfg["target_col"],
    protected_attributes=hc_cfg["protected_attributes"],
    figures_dir=CONFIG["paths"]["figures_dir"],
    report_path=CONFIG["paths"]["metrics_dir"] / "eda_home_credit.md",
    dataset_name="home_credit",
)
stats
```

See `reports/figures/home_credit/*.png` for all generated plots (target distribution, missingness, numeric distributions, correlation heatmap, target rate by `CODE_GENDER` / `AGE_GROUP`).
