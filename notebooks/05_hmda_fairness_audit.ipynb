# 05 — HMDA Fairness-Only Audit

HMDA records mortgage **origination decisions** (approved/denied), not loan
performance — it is **not** used to build a second "default" model, and its
role here is strictly to test whether the fairness-auditing methodology from
notebook 04 generalizes to a second, independently-labeled dataset and
decision task (see `src/dac/fairness/hmda_audit.py` docstring, and
Cross-Question 1 in the mentor/reviewer critique).

Protected attributes (`derived_race`, `derived_sex`) are excluded from the
model's features and used only for the post-hoc audit.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent / "src"))

from dac.config import CONFIG
from dac.data.loader import load_hmda
from dac.eda.eda_report import run_eda
from dac.fairness.hmda_audit import run_hmda_fairness_audit

hmda_cfg = CONFIG["data"]["hmda"]
df, is_synthetic = load_hmda()
print(f"Loaded {df.shape[0]:,} rows x {df.shape[1]} cols | synthetic={is_synthetic}")
df.head()
```

```python
eda_stats = run_eda(
    df,
    target_col=hmda_cfg["target_col"],
    protected_attributes=hmda_cfg["protected_attributes"],
    figures_dir=CONFIG["paths"]["figures_dir"],
    report_path=CONFIG["paths"]["metrics_dir"] / "eda_hmda.md",
    dataset_name="hmda",
)
eda_stats
```

```python
results = run_hmda_fairness_audit(
    df,
    target_col=hmda_cfg["target_col"],
    id_col=hmda_cfg["id_col"],
    protected_attributes=hmda_cfg["protected_attributes"],
    figures_dir=CONFIG["paths"]["figures_dir"] / "hmda",
    metrics_dir=CONFIG["paths"]["metrics_dir"] / "fairness",
    seed=CONFIG["seed"],
)
results["performance"]
```
